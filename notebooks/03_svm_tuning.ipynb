# 03 — SVM tuning
Grid search over kernel, C, gamma/degree, and number of PCA components.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))  # so `import src...` works from notebooks/

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src.data import load_train, load_test, to_grayscale, to_images
from sklearn.model_selection import GridSearchCV
from src.models import svm_pipeline, svm_param_grid
from src.evaluate import cv_splitter
```

```python
ids, X_raw, y = load_train()
X = to_grayscale(X_raw)
```

```python
search = GridSearchCV(svm_pipeline(), svm_param_grid(), cv=cv_splitter(), scoring='roc_auc', n_jobs=-1)
search.fit(X, y)
print(search.best_score_, search.best_params_)
```

```python
results = pd.DataFrame(search.cv_results_)
cols = ['param_clf__kernel', 'param_clf__C', 'param_clf__gamma', 'param_pca__n_components', 'mean_test_score', 'std_test_score']
results.sort_values('mean_test_score', ascending=False)[cols].head(15)
```
